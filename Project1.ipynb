**3. Data Visualization:**

Data visualization of the total population:

```python
plt.figure(figsize=(8,5))
plt.hist(df["Total.Population"], bins=15, color="blue", edgecolor="black")
plt.title("Distribution of Total Population")
plt.xlabel("Total Population")
plt.ylabel("Number of States")
plt.show()
```

This histogram shows how population sizes vary across U.S. states. Most states have moderate population sizes, while a few states have extremely high populations. The right-skewed shape suggests that a small number of highly populated states (like California, Texas, and Florida) significantly increase the overall range.

Relationship between total number of cases and population:

```python
plt.figure(figsize=(8,5))
plt.scatter(df["Total.Population"], df["Total.Number"], color="coral")
plt.title("Relationship Between Total Cancer Cases and State Population")
plt.xlabel("Total Population")
plt.ylabel("Total Number of Cancer Cases")
plt.show()
```

This scatter plot shows that states with higher populations tend to have a greater number of total cancer cases. The upward trend indicates a strong positive relationship between population size and cancer cases. This suggests that population is a major factor influencing raw case numbers, meaning comparisons between states should consider population size or use rate-based measures instead of totals.
