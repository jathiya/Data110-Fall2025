```python
from matplotlib import pyplot as plt
import numpy as np
import pandas as pd
plt.style.use("ggplot")
```

Task 1: Basic Bar Chart for Crime Rates in 2023
In this task, you will create a bar chart showing the offense rates per 100,000 people for each type of crime in 2023.

Instructions:
Create a bar chart for crime rates in 2023.
Customize the bar colors and add gridlines.
Label the axes and add a title.

```python
data = {'Offense': ['Aggravated Assault','Burglary','Larceny','Motor Vehicle Theft','Murder','Rape (revised definition)','Robbery'],
    'Rate_2023': [264.1, 250.7, 1347.2, 318.7, 5.7, 38.0, 66.5]}

df = pd.DataFrame(data)

plt.figure(figsize=(10,6))
plt.bar(
    df['Offense'],
    df['Rate_2023'],
    color=['yellow', 'orange', 'green', 'red', 'purple', 'brown', 'pink']
)
plt.xlabel("Offense Type")
plt.ylabel("Offenses per 100,000 People (2023)")
plt.title("Crime Rates per 100,000 People in 2023")

plt.grid(axis="y", linestyle="--")
plt.tight_layout()
plt.show()
```

```python
offenses = ['Aggravated Assault', 'Burglary', 'Larceny',
            'Motor Vehicle Theft', 'Murder', 'Rape (revised definition)', 'Robbery']
changes = [-3.3, -8.1, -4.9, 12.0, -12.0, -9.8, -0.8]

plt.figure(figsize=(10, 6))
plt.barh(offenses, changes, color=['yellow', 'orange', 'green', 'red', 'purple', 'brown', 'pink'])

plt.title('Year to Year Changes in Crime Rates')
plt.xlabel('Change in Offenses per 100,000 (2022 to 2023)')
plt.ylabel('Offense Type')

plt.grid(axis='x', linestyle='--')

plt.tight_layout()
plt.show()
```

Optional Task: Combined Bar Chart for Crime Rates in 2022 and 2023
In this optional task, you will create a combined bar chart that shows the crime rates for both 2022 and 2023.

Instructions:
Calculate the crime rates for 2022 based on the 2023 values and the percentage changes provided.
Create a combined bar chart that compares the crime rates for 2022 and 2023 for each offense type.
Use different colors to distinguish between the years 2022 and 2023.
Label the axes and add a title.
Add a legend to indicate which bars represent which year.

```python
offenses = ['Aggravated Assault', 'Burglary', 'Larceny',
            'Motor Vehicle Theft', 'Murder', 'Rape (revised definition)', 'Robbery']
rates_2023 = [264.1, 250.7, 1347.2, 318.7, 5.7, 38.0, 66.5]
rates_2022 = [267.4, 258.8, 1352.1, 306.7, 17.7, 47.8, 67.3]

plt.figure(figsize=(10,6))
plt.bar(offenses, rates_2022, label='2022', color='skyblue')
plt.bar(offenses, rates_2023, label='2023', color='purple', bottom=rates_2022)

plt.xlabel("Offense Type")
plt.ylabel("Offenses per 100,000 People")
plt.title("Crime Rates per 100,000: 2022 vs 2023")

plt.grid(axis='y', linestyle='--')
plt.legend()

plt.tight_layout()
plt.show()
```
