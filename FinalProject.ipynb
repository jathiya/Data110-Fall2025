Climate change does not affect all communities equally. Some populations are more vulnerable due to socioeconomic, environmental, and health related factors. This project will explor climate vulnerability across Maryland counties to identify patterns and relationships between vulnerability scores and contributing factors. Using exploratory data analysis and data visualizations, this study aims to tell a data driven story.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px

df = pd.read_csv('/content/MD_Climate_Vulnerability_Score (1) (1).csv')
df.head()
```

I began by loading the dataset and looking at its structure, data types, and summary statistics to understand the variables.

## Exploratory Data Analysis (EDA)

The exploratory data analysis focuses on understanding how climate vulnerability varies across Maryland counties and identifying which factors are most strongly associated with higher vulnerability scores. Key questions include:

- Which counties have the highest and lowest climate vulnerability?
- How does vulnerability relate to socioeconomic and environmental indicators?
- Are there visible geographic patterns in vulnerability across the state?


```python
plt.figure(figsize=(8,5))
sns.boxplot(x='FLOOD_SS', y='CVS_RAW', data=df)
plt.title('Climate Vulnerability by Flood Susceptibility Level')
plt.xlabel('Flood Susceptibility')
plt.ylabel('Climate Vulnerability Score (Raw)')
plt.show()
```

This boxplot shows how climate vulnerability scores vary across different flood susceptibility levels. Counties with higher flood susceptibility generally have higher median vulnerability scores.




```python
plt.figure()
plt.scatter(df['UHI'], df['CVS_RAW'])
plt.xlabel('Urban Heat Island Index')
plt.ylabel('Climate Vulnerability Score')
plt.title('Relationship Between Urban Heat and Climate Vulnerability')
plt.show()
```

This scatter plot shows a positive relationship between urban heat island intensity and climate vulnerability, indicating that counties experiencing higher heat exposure also tend to face greater overall climate risk.

```python
plt.figure(figsize=(8,5))
plt.hist(df['CVS_RAW'], bins=20, edgecolor='black')
plt.title('Distribution of Climate Vulnerability Scores in Maryland')
plt.xlabel('Climate Vulnerability Score (Raw)')
plt.ylabel('Number of Counties')
plt.show()
```

The distribution of climate vulnerability scores shows a broad spread across Maryland counties, with most counties clustered around moderate vulnerability levels and fewer counties experiencing low or high scores.

```python
fig1 = px.scatter(df, x='P_UNDERSER',
    y='CVS_RAW',
    hover_data=['GEOID20'],
    title='Climate Vulnerability vs Underserved Populations',
    labels={'P_UNDERSER': 'Percent Underserved Population', 'CVS_RAW': 'Climate Vulnerability Score (Raw)'})

fig1.show()
```

This interactive scatter plot shows how climate vulnerability changes as the percentage of underserved populations increases. Counties with more underserved residents have higher climate vulnerability, showing that climate risk is closely connected to social inequality.

```python
fig = px.scatter(
    df,
    x='FLOOD_SS',
    y='CVS_RAW',
    hover_data=['GEOID20'],
    title='Climate Vulnerability vs Flood Susceptibility',
    labels={'FLOOD_SS': 'Flood Susceptibility', 'CVS_RAW': 'Climate Vulnerability Score (Raw)'})
fig.show()

```

This interactive scatter plot shows the relationship between flood susceptibility and overall climate vulnerability. Counties with higher flood risk often have higher climate vulnerability, and the interactive features allow closer comparison of individual counties.

This project analyzed climate vulnerability across Maryland counties using data visualizations. The results show that climate vulnerability varies across the state and is often higher in counties facing greater social and environmental risks. Overall, the findings suggest that climate vulnerability is closely linked to both environmental exposure and social factors.
