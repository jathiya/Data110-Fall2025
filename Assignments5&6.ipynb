```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/jathiya/Data110-21843--fall25/refs/heads/main/dataset/titanic.csv"
titanic = pd.read_csv(url)

titanic.head()
```

Understanding the dataset:
The Titanic dataset contains information about passengers on the Titanic. Which includes demographics and survival outcomes. I will be focusing on the variables survival status, passenger class, sex, age, fare, town, and the number of siblings, spouses, parents, and children on the ship. These variables are important for understanding who had higher chances of survival.

```python
import seaborn as sns
titanic = sns.load_dataset("titanic")
titanic.head()
```

```python
sns.histplot(titanic["fare"], bins=30, kde=True)
plt.title("Distribution of Fares on the Titanic")
plt.xlabel("Fares")
plt.ylabel("Amount")
plt.show()
```

The fare distribution is strongly right skewed, which means most passengers paid low fares, and few paid high ones. These high fare outliers are likely due to first class travelers. This pattern reflects clear socioeconomic inequalities on the ship.

Task 2:

```python
sns.kdeplot(data=titanic, x="fare")
plt.title("Fare Distribution by Survival Status")
plt.xlabel("Fare")
plt.show()
```

Based on the data presented, most survivors had higher fares on average, showing that passengers in wealthier classes had better access to lifeboats. The ones who didnt survive's distribution peaks at lower fares, showing the link between class and survival.

Task 3:

```python
sns.boxplot(data=titanic, x="class", y="fare", hue='sex')
plt.title("Fare Distribution by Class and Sex")
plt.show()
```

Regardless of sex, first class passengers paid the most, while third-class fares were the lowest. Males and females within each class show similar medians but male fares may have a bit more outliers.

Task 5:

```python
sns.violinplot(data=titanic, x="sex", y="age")
plt.title("Age Distribution by Sex")
plt.show()
```

The age distribution is similar of men and women. But females show a slightly higher area around young adult ages, while males have more spread across all ages, including children and older adults.

Task 6:

```python
sns.violinplot(data=titanic, x="sex", y="age")
sns.stripplot(data=titanic, x="sex", y="age", color="black")
plt.title("Age Distribution with Individual Data Points")
plt.show()
```

Adding each individual point shows where the data actually clusters, for example many passengers in their 20s and 30s. This also highlights outliers that the smooth violin shape hides.
